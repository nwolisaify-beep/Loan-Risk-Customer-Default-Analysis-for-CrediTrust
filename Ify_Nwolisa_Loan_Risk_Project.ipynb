Remove Extreme Outliers Using IQR
¶

```python
Q1 = df['Loan_Amount'].quantile(0.25)
Q3 = df['Loan_Amount'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[
    (df['Loan_Amount'] >= lower_bound) &
    (df['Loan_Amount'] <= upper_bound)
]
```

Task 4: Customer Segmentation
¶

Task (4.1) Group customers into High Risk, Medium Risk, Low Risk
¶

Task (4.2) Explain your logic clearly
¶

Create Installment Burden Ratio
¶

```python
df['Installment_Burden'] = (
    df['Monthly_Installment'] / df['Income']
)
```

Create Risk Categories
¶

```python
df['Installment_Burden'] = (
    df['Monthly_Installment'] / df['Income']
)
```

```python
def risk_category(row):

    if (
        row['Credit_Score'] < 500 or
        row['Previous_Default'] == 'Yes' or
        row['Employment_Status'] == 'Unemployed' or
        row['Installment_Burden'] > 0.35
    ):

        return 'High Risk'

    elif 500 <= row['Credit_Score'] <= 650:

        return 'Medium Risk'

    else:

        return 'Low Risk'
```

Apply Segmentation
¶

```python
df['Risk_Category'] = df.apply(
    risk_category,
    axis=1
)
```

View Segmentation Results
¶

```python
print(df['Risk_Category'].value_counts())
```

Visualization
¶

```python
sns.countplot(
    x='Risk_Category',
    data=df
)

plt.title('Customer Risk Segmentation')

plt.show()
```

Task 6: Optional (Advanced)
¶

Task (6.1) Build a simple model to predict default
¶

Create Approval Rules
¶

```python
def loan_decision(row):

    if row['Credit_Score'] < 500:
        return 'Reject'

    elif row['Previous_Default'] == 'Yes':
        return 'Manual Review'

    elif row['Installment_Burden'] > 0.35:
        return 'Reduce Loan Amount'

    elif row['Employment_Status'] == 'Unemployed':
        return 'Strict Verification'

    else:
        return 'Approve'
```

Apply Approval Rules
¶

```python
df['Loan_Decision'] = df.apply(
    loan_decision,
    axis=1
)
```

```python
print(df['Loan_Decision'].value_counts())
```
